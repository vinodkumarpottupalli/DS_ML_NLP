```python
from datetime import datetime
from pydantic import BaseModel
```

```python
class Delivery(BaseModel):
    timestamp: datetime
    dimensions: tuple[int, int]
```

```python
m = Delivery(timestamp='2020-01-02T03:04:05Z', dimensions=['10', '20'])
```

```python
m = Delivery(timestamp='2020-01-02T03:04:05Z', dimensions=[10,11])
```

```python
m = Delivery(timestamp='2020-01-02T03:04:05Z', dimensions=[10,])
```

```python
m = Delivery(timestamp='2020-01-02T03:04:05Z', dimensions=['hello', 'world'])
```
